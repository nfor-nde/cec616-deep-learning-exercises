## Section A — Artificial Neural Networks (ANN) — Q1–Q15

**Q1.** What is the basic computational unit of an ANN called?
**Answer: A — Neuron / Perceptron.** The neuron (perceptron) is the fundamental processing unit that receives inputs, applies weights/bias, and passes the result through an activation function.

**Q2.** Which function introduces non-linearity into a neural network?
**Answer: B — Activation function.** Without a non-linear activation function, stacking layers would collapse into a single linear transformation, preventing the network from learning complex patterns.

**Q3.** What does the bias term in a neuron do?
**Answer: B — Shifts the activation function.** The bias allows the activation curve to shift left/right, giving the model flexibility to fit data that doesn't pass through the origin.

**Q4.** Which algorithm computes gradients for updating weights?
**Answer: B — Backpropagation.** It applies the chain rule to propagate the error gradient backward from the output layer to earlier layers.

**Q5.** The vanishing gradient problem is most associated with which activation function?
**Answer: B — Sigmoid.** Its derivative saturates near 0 for large |z|, so gradients shrink exponentially across many layers during backpropagation.

**Q6.** Purpose of the softmax activation function?
**Answer: B — Convert logits to a probability distribution over classes.** Softmax exponentiates and normalizes outputs so they sum to 1, ideal for multi-class classification.

**Q7.** Which optimizer uses adaptive learning rates?
**Answer: B — Adam.** Adam combines momentum with per-parameter adaptive learning rates (using estimates of first and second moments of gradients).

**Q8.** What is "overfitting"?
**Answer: B — Model performs well on training data but poorly on unseen data.** The model has memorized noise/specifics of the training set rather than learning generalizable patterns.

**Q9.** Dropout is used to:
**Answer: B — Prevent overfitting by randomly disabling neurons during training.** This forces the network to not rely on any single neuron, improving generalization.

**Q10.** Loss function typically used for multi-class classification?
**Answer: B — Categorical Cross-Entropy.** It measures the divergence between the predicted probability distribution and the true one-hot label.

**Q11.** What is an epoch?
**Answer: B — One complete pass through the entire training dataset.**

**Q12.** Which layer connects every neuron to every neuron in the next layer?
**Answer: C — Fully connected (Dense) layer.**

**Q13.** Batch normalization helps primarily by:
**Answer: A — Reducing internal covariate shift and stabilizing training.** It normalizes layer inputs, which speeds up and stabilizes convergence.

**Q14.** What does "learning rate" control?
**Answer: B — Size of the steps taken towards minimizing the loss.**

**Q15.** Which is NOT a common activation function?
**Answer: D — Gradient Descent.** Gradient Descent is an optimization algorithm, not an activation function (ReLU, Tanh, Sigmoid all are).

### 1. Differentiate between Machine Learning and Deep Learning

| Aspect | Machine Learning (ML) | Deep Learning (DL) |
|---|---|---|
| Feature engineering | Manual — features hand-crafted by domain experts | Automatic — features learned from raw data by the network itself |
| Data requirement | Works reasonably with smaller datasets | Needs large amounts of data to perform well |
| Compute requirement | Lower (can run on CPU) | High — usually needs GPUs/TPUs |
| Model structure | Algorithms like Decision Trees, SVM, k-NN, Linear/Logistic Regression | Multi-layer neural networks (CNN, RNN, Transformer, etc.) |
| Interpretability | Generally more interpretable | Often a "black box", harder to interpret |
| Example | Predicting house price with Linear Regression | Recognising objects in an image with a CNN |

In short: **Deep Learning is a subset of Machine Learning** that uses deep (multi-layered) neural networks to automatically learn hierarchical feature representations directly from raw data.

### 4. General equation for Multivariate Linear Regression with 4 independent variables

$$
\hat{y} = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + \theta_3 x_3 + \theta_4 x_4
$$

where:
- $\hat{y}$ is the predicted output,
- $\theta_0$ is the intercept (bias),
- $\theta_1, \theta_2, \theta_3, \theta_4$ are the coefficients (weights) for the four independent variables $x_1, x_2, x_3, x_4$.

In vector form: $\hat{y} = \boldsymbol{\theta}^T \mathbf{x}$, with $\mathbf{x} = [1, x_1, x_2, x_3, x_4]^T$.


### 5. General equation of Logistic Regression + sigmoid plot

Logistic Regression models the probability of the positive class using the **sigmoid (logistic) function** applied to a linear combination of inputs:

$$
z = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + \dots + \theta_n x_n
$$

$$
h_\theta(x) = \sigma(z) = \frac{1}{1 + e^{-z}}
$$

The output $h_\theta(x) \in (0, 1)$ is interpreted as $P(y=1 \mid x)$.


````python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-10, 10, 200)
sigmoid = 1 / (1 + np.exp(-z))

plt.figure(figsize=(6,4))
plt.plot(z, sigmoid, color='darkorange', linewidth=2)
plt.axhline(0.5, color='gray', linestyle='--', linewidth=1)
plt.axvline(0, color='gray', linestyle='--', linewidth=1)
plt.title('Sigmoid Function: $\\sigma(z) = 1 / (1 + e^{-z})$')
plt.xlabel('z')
plt.ylabel('$\\sigma(z)$')
plt.grid(alpha=0.3)
plt.show()
````

### 6. Purpose of splitting the dataset into training and testing sets

The dataset is split so that the model can be evaluated on data it has **never seen during training**. This:
- Gives an honest estimate of how the model will generalize to new, real-world data.
- Helps detect **overfitting** (a model that memorizes training data will score much worse on the held-out test set).
- Provides a fair basis for comparing different models/hyperparameters.

A common split is 70–80% training / 20–30% testing (sometimes a third **validation** split is also used for hyperparameter tuning).


### 7. Differentiate between Bias and Variance

| | Bias | Variance |
|---|---|---|
| Definition | Error from overly simplistic assumptions in the model | Error from excessive sensitivity to small fluctuations in the training data |
| Symptom | **Underfitting** — poor performance on both train and test data | **Overfitting** — very good on training data, poor on test data |
| Cause | Model too simple (e.g., linear model for non-linear data) | Model too complex (e.g., very deep tree with no pruning) |
| Fix | Use a more complex/expressive model, add features | Regularization, more data, simpler model, dropout, cross-validation |

The **bias–variance tradeoff** states that decreasing bias tends to increase variance and vice versa; the goal is to find the sweet spot that minimizes total generalization error.


### 8. Differentiate between Decision Tree and Random Forest

| | Decision Tree | Random Forest |
|---|---|---|
| Structure | A single tree of if/else splits | An ensemble (bagging) of many decision trees |
| Variance | High variance — prone to overfitting | Lower variance — averaging/voting across trees reduces overfitting |
| Training | Fast, simple to interpret | Slower (many trees), but more robust |
| Prediction | One tree's output | Majority vote (classification) or average (regression) of all trees |
| Feature use | Uses all features at each split | Each tree uses a random subset of features (and a bootstrap sample of data), increasing diversity |

In short, a **Random Forest is an ensemble of Decision Trees** that improves accuracy and reduces overfitting through bagging and feature randomness.


### 9. Formulae for Accuracy and Recall

$$
\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}
$$

$$
\text{Recall (Sensitivity / TPR)} = \frac{TP}{TP + FN}
$$

where $TP$ = True Positives, $TN$ = True Negatives, $FP$ = False Positives, $FN$ = False Negatives.


## Confusion Matrix Question

Given confusion matrix:

| | Predicted Positive | Predicted Negative |
|---|---|---|
| **Actual Positive** | 85 | 12 |
| **Actual Negative** | 10 | 75 |

### 1. Identify TP, FP, TN, FN
- **TP (True Positive)** = 85 — actual positive, predicted positive
- **FN (False Negative)** = 12 — actual positive, predicted negative
- **FP (False Positive)** = 10 — actual negative, predicted positive
- **TN (True Negative)** = 75 — actual negative, predicted negative

### 2. Formulas and calculations


````python
TP, FN, FP, TN = 85, 12, 10, 75

# a. False Positive Rate, TPR (Recall/Sensitivity), TNR (Specificity)
FPR = FP / (FP + TN)
TPR = TP / (TP + FN)
TNR = TN / (TN + FP)

# b. Sensitivity (same as TPR), Precision
Sensitivity = TPR
Precision = TP / (TP + FP)

# c. F1-score, Accuracy
F1 = 2 * (Precision * Sensitivity) / (Precision + Sensitivity)
Accuracy = (TP + TN) / (TP + TN + FP + FN)

print(f"False Positive Rate (FPR) = FP/(FP+TN) = {FP}/{FP+TN} = {FPR:.4f}")
print(f"True Positive Rate  (TPR) = TP/(TP+FN) = {TP}/{TP+FN} = {TPR:.4f}")
print(f"True Negative Rate  (TNR) = TN/(TN+FP) = {TN}/{TN+FP} = {TNR:.4f}")
print(f"Sensitivity               = {Sensitivity:.4f}  (same formula as TPR)")
print(f"Precision = TP/(TP+FP)    = {TP}/{TP+FP} = {Precision:.4f}")
print(f"F1-score  = 2*P*R/(P+R)   = {F1:.4f}")
print(f"Accuracy  = (TP+TN)/Total = {Accuracy:.4f}")
````

**Summary of results:**

| Metric | Formula | Value |
|---|---|---|
| False Positive Rate | FP / (FP+TN) | 10/85 ≈ **0.1176** |
| TPR (Recall) | TP / (TP+FN) | 85/97 ≈ **0.8763** |
| TNR (Specificity) | TN / (TN+FP) | 75/85 ≈ **0.8824** |
| Sensitivity | = TPR | ≈ **0.8763** |
| Precision | TP / (TP+FP) | 85/95 ≈ **0.8947** |
| F1-score | 2·P·R / (P+R) | ≈ **0.8854** |
| Accuracy | (TP+TN) / Total | 160/182 ≈ **0.8791** |

### Question — Reinforcement Learning State Diagram (Q-Learning)

The state diagram shows 6 rooms/states (0–5) with state **5 = Goal**. The connections (edges) and their immediate rewards read from the diagram are:

- 0 ↔ 4 : reward 0
- 1 ↔ 3 : reward 0
- 1 ↔ 5 : reward **100** (leads to goal)
- 2 ↔ 3 : reward 0
- 3 ↔ 4 : reward 0
- 3 ↔ 5 : reward **100** (leads to goal)
- 4 ↔ 5 : reward **100** (leads to goal)
- 5 ↔ 5 : reward 100 (staying at the goal)

#### 1. R Matrix

Rows = current state, Columns = action (move to that state). `-1` (or `-`) means "no direct connection / action not available".


````python
import numpy as np

#        0    1    2    3    4    5
R = np.array([
    [-1, -1, -1, -1,  0, -1],   # state 0
    [-1, -1, -1,  0, -1, 100],  # state 1
    [-1, -1, -1,  0, -1, -1],   # state 2
    [-1,  0,  0, -1,  0, -1],   # state 3
    [ 0, -1, -1,  0, -1, 100],  # state 4
    [-1,  0, -1, -1,  0, 100],  # state 5 (goal, can stay)
])

print("R Matrix:")
print(R)
````

#### 2. General Q(State, Action) equation (Bellman equation for Q-Learning)

$$
Q(s, a) = R(s, a) + \gamma \max_{a'} Q(s', a')
$$

where:
- $R(s,a)$ is the immediate reward for taking action $a$ in state $s$,
- $s'$ is the resulting next state,
- $\gamma$ is the discount factor ($0 \le \gamma \le 1$),
- $\max_{a'} Q(s', a')$ is the best achievable Q-value from the next state.

(With a learning rate $\alpha$, the *incremental* update form used during training is:
$Q(s,a) \leftarrow Q(s,a) + \alpha\big[R(s,a) + \gamma \max_{a'} Q(s',a') - Q(s,a)\big]$.)

#### 3. Calculate Q(4, 5) with γ = 0.4

Assume the Q-table is initialized to all zeros (standard initial condition for this classic example).

$$
Q(4,5) = R(4,5) + \gamma \max_{a'} Q(5, a')
$$

Since all $Q(5, a')$ values are still 0 (untrained table), $\max_{a'} Q(5,a') = 0$:

$$
Q(4,5) = 100 + 0.4 \times 0 = \mathbf{100}
$$


````python
gamma = 0.4
Q = np.zeros_like(R, dtype=float)  # Q-table initialized to zero

R_4_5 = R[4, 5]                 # immediate reward for action 5 from state 4
max_Q_next = Q[5, :].max()      # best Q-value achievable from state 5 (all zero initially)

Q_4_5 = R_4_5 + gamma * max_Q_next
print(f"R(4,5) = {R_4_5}")
print(f"max_a' Q(5,a') = {max_Q_next}")
print(f"Q(4,5) = {R_4_5} + {gamma} * {max_Q_next} = {Q_4_5}")
````

### Question — Perceptron with sigmoid activation (weights: bias = −10, $x_1$ = 20, $x_2$ = 20)

#### 1. $h_\theta(x)$ equation

$$
h_\theta(x) = \sigma(-10 + 20x_1 + 20x_2) = \frac{1}{1+e^{-(-10+20x_1+20x_2)}}
$$

#### 2. Truth table (binary inputs)


````python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

w0, w1, w2 = -10, 20, 20  # bias, weight on x1, weight on x2

print(f"{'x1':>3} {'x2':>3} {'z':>6} {'h(x)':>8} {'rounded'}")
for x1 in [0, 1]:
    for x2 in [0, 1]:
        z = w0 + w1*x1 + w2*x2
        h = sigmoid(z)
        print(f"{x1:>3} {x2:>3} {z:>6} {h:>8.4f} {round(h)}")
````

### Question — ANN Diagram ($x_1, x_2, x_m \rightarrow z_1, z_2 \rightarrow y_1, y_2$)

#### 4. Annotate the various weights

Each input $x_j$ connects to each node $z_i$ with its own weight $w_{ji}$ (weight from input $j$ to unit $i$), plus a bias term $b_i$ for each unit $z_i$:

- $x_1 \to z_1$: weight $w_{11}$ &nbsp;&nbsp; $x_1 \to z_2$: weight $w_{12}$
- $x_2 \to z_1$: weight $w_{21}$ &nbsp;&nbsp; $x_2 \to z_2$: weight $w_{22}$
- $x_m \to z_1$: weight $w_{m1}$ &nbsp;&nbsp; $x_m \to z_2$: weight $w_{m2}$
- Bias units: $b_1$ (into $z_1$), $b_2$ (into $z_2$)

#### 5. Equation for $z_1$

$$
z_1 = w_{11}x_1 + w_{21}x_2 + \dots + w_{m1}x_m + b_1
$$

#### 6. Equation for $y_1$ (sigmoid activation)

$$
y_1 = g(z_1) = \sigma(z_1) = \frac{1}{1+e^{-z_1}}
$$

#### 7. Generalised form of $z_i$

$$
z_i = \sum_{j=1}^{m} w_{ji}x_j + b_i = \mathbf{w}_i^T \mathbf{x} + b_i
$$

#### 8. The Chain Rule principle

The chain rule states that the derivative of a composite function is the product of the derivatives of its parts. In a neural network, the loss $L$ depends on $y_1$, which depends on $z_1$, which depends on the weights $w_{j1}$. The chain rule lets us compute how the loss changes with respect to an early-layer weight by **multiplying the local derivatives along the computational path**:

$$
\frac{\partial L}{\partial w_{j1}} = \frac{\partial L}{\partial y_1} \cdot \frac{\partial y_1}{\partial z_1} \cdot \frac{\partial z_1}{\partial w_{j1}}
$$

This is the mathematical foundation of **backpropagation**: gradients are propagated backward from the output layer to earlier layers by repeatedly applying the chain rule, layer by layer.

### Topic 10 (bis) — Q-Learning Model for Labyrinth Navigation

**Data preparation:** No labeled dataset — define the **environment** instead: grid/maze layout, valid moves, walls, start cell, and goal cell with associated rewards (e.g., +100 at goal, −1 per step, large negative for hitting a wall).

**Architecture/Pipeline:** `Environment (maze) ⇄ Agent` loop: `observe state → choose action (ε-greedy) → execute action → receive reward + next state → update Q-table via Bellman equation → repeat until terminal state`

**Model-building steps:** 1) Encode the maze as states (cells) and actions (up/down/left/right); 2) Initialize Q-table to zeros; 3) For each episode, run ε-greedy action selection and update Q-values using the Bellman update; 4) Decay ε over episodes; 5) After training, extract the optimal policy by taking `argmax_a Q(s,a)` at each state.

**Key pseudocode:**
```
Q = np.zeros((n_states, n_actions))
alpha, gamma, epsilon = 0.1, 0.9, 1.0

for episode in range(n_episodes):
    s = start_state
    while s != goal_state:
        if random() < epsilon:
            a = random_action()
        else:
            a = np.argmax(Q[s])
        s_next, r = env.step(s, a)
        Q[s,a] += alpha * (r + gamma * np.max(Q[s_next]) - Q[s,a])
        s = s_next
    epsilon = max(0.01, epsilon * 0.995)   # decay exploration

policy = np.argmax(Q, axis=1)   # optimal action per state
```

**Key evaluation metrics:** Average reward per episode (learning curve), number of steps to reach goal over time, success rate, convergence of the Q-table.

````python
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
import matplotlib.pyplot as plt
import numpy as np

# ---- 1. Data preparation ----
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()
class_names = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']

X_train = X_train.astype('float32') / 255.0
X_test  = X_test.astype('float32') / 255.0
y_train = y_train.flatten()
y_test  = y_test.flatten()

print("Train shape:", X_train.shape, " Test shape:", X_test.shape)
````

````python
# ---- 2. Build the CNN architecture ----
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.Conv2D(32, (3,3), activation='relu', padding='same'),
    layers.MaxPooling2D((2,2)),
    layers.Dropout(0.25),

    layers.Conv2D(64, (3,3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.Conv2D(64, (3,3), activation='relu', padding='same'),
    layers.MaxPooling2D((2,2)),
    layers.Dropout(0.25),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
````

````python
# ---- 3. Set callbacks to control the training phase ----
early_stop = callbacks.EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True)
checkpoint = callbacks.ModelCheckpoint('best_cifar_model.keras', monitor='val_accuracy', save_best_only=True)
reduce_lr  = callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-6)

# ---- 4. Train the model ----
history = model.fit(
    X_train, y_train,
    validation_split=0.1,
    epochs=2,            # increase for higher accuracy
    batch_size=64,
    callbacks=[early_stop, checkpoint, reduce_lr],
    verbose=1
)
````

````python
# ---- 5. Plot LOSS and ACCURACY for training and validation ----
fig, axes = plt.subplots(1, 2, figsize=(12,4))

axes[0].plot(history.history['loss'], label='Train Loss')
axes[0].plot(history.history['val_loss'], label='Val Loss')
axes[0].set_title('Loss over Epochs')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Loss')
axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(history.history['accuracy'], label='Train Accuracy')
axes[1].plot(history.history['val_accuracy'], label='Val Accuracy')
axes[1].set_title('Accuracy over Epochs')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Accuracy')
axes[1].legend(); axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

````python
# ---- 6. Evaluate the score on the test set ----
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")

from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

y_pred = np.argmax(model.predict(X_test), axis=1)
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=class_names))

cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
fig, ax = plt.subplots(figsize=(8,8))
disp.plot(ax=ax, cmap='Blues', xticks_rotation=45)
plt.title("CIFAR-10 Test Set Confusion Matrix")
plt.show()
````

````python
# ---- 7. Demo: show predictions on a few test images ----
fig, axes = plt.subplots(2, 5, figsize=(12,5))
idxs = np.random.choice(len(X_test), 10, replace=False)

for ax, idx in zip(axes.flat, idxs):
    ax.imshow(X_test[idx])
    true_label = class_names[y_test[idx]]
    pred_label = class_names[y_pred[idx]]
    color = 'green' if true_label == pred_label else 'red'
    ax.set_title(f"T:{true_label}\nP:{pred_label}", color=color, fontsize=9)
    ax.axis('off')

plt.suptitle("CIFAR-10 — Sample Predictions (Demo)")
plt.tight_layout()
plt.show()
````
